# 2. Baseline Model Training

## I. Split A (Leakage-safe)

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

train_df = pd.read_csv("../data/processed/train.csv")
test_df  = pd.read_csv("../data/processed/test.csv")

# Separating the data into X (prompt text) and y (isInjection = 1/0)
X_train, y_train = train_df["text"], train_df["label"].values
X_test,  y_test  = test_df["text"],  test_df["label"].values

vec_safe = TfidfVectorizer(max_features=10000, ngram_range=(1,2))
X_train_vec = vec_safe.fit_transform(X_train)
X_test_vec  = vec_safe.transform(X_test)
```

### Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

logModel_a = LogisticRegression(max_iter=1000).fit(X_train_vec, y_train)
preds = logModel_a.predict(X_test_vec)
print(classification_report(y_test,preds))
```

### Random Forest Classifier
- Random state of 50 to get consistent results

```python
from sklearn.ensemble import RandomForestClassifier

rfModel_a = RandomForestClassifier(random_state=50).fit(X_train_vec, y_train)
preds = rfModel_a.predict(X_test_vec)
print(classification_report(y_test,preds))
```

### Decision Tree Classifier
- Random state of 50 to get consistent results

```python
from sklearn.tree import DecisionTreeClassifier

dtModel_a = DecisionTreeClassifier(random_state=50).fit(X_train_vec, y_train)
preds = dtModel_a.predict(X_test_vec)
print(classification_report(y_test,preds))
```

## II. Split B (Naive)
